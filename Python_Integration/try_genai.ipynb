```python
from dotenv import load_dotenv
from langchain_community.chat_models import ChatSnowflakeCortex
from langchain.messages import HumanMessage, SystemMessage
import os
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate, SystemMessagePromptTemplate,HumanMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv()
```

```python
chatModel=ChatSnowflakeCortex(
    model='mistral-large',
    cortex_function='complete',
    temperature=0.2,
    max_tokens=100,
    top_p=0.95,
    account=os.environ.get("SNOWFLAKE_ACCOUNT"),
    username=os.environ.get("SNOWFLAKE_USER"),
    password=os.environ.get("SNOWFLAKE_PASSWORD"),
    database=os.environ.get("SNOWFLAKE_DATABASE"),
    schema=os.environ.get("SNOWFLAKE_SCHEMA"),
    role=os.environ.get('SNOWFLAKE_ROLE'),
    warehouse=os.environ.get('SNOWFLAKE_WAREHOUSE')
)
```

```python
messages=ChatPromptTemplate.from_messages(
    messages=[
        SystemMessagePromptTemplate.from_template(template="You are a friendly assistant"),
        HumanMessagePromptTemplate.from_template(template="Answer topic related to : {topic}")
    ]
)
```

```python
outputParser=StrOutputParser()
```

```python
chain=messages|chatModel|outputParser
```

```python
chain.invoke(input={'topic':"Cricket"})
```
