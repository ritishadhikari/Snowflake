```
import pandas as pd

# We can also use Snowpark for our analyses!
from snowflake.snowpark.context import get_active_session
from snowflake.ml.registry import Registry
```

```
session = get_active_session()
```

```
session.query_tag="model-registry-2"
```

```
registry=Registry(session=session)
```

```
registry.show_models()
```

```
registry.show_models()['name']
```

```
lastModel=pd.DataFrame(data=registry.show_models()).\
    sort_values('created_on',ascending=False).\
    loc[:1,'name'].values[0]
lastModel
```

```
model=registry.get_model(lastModel)
```

```
print(f"Model Name: {model.name}")
print(f"Fully Qualified Name: {model.fully_qualified_name}")
print(f"Description: {model.description}")
print(f"Comment: {model.comment}")

print(f"Tags: {model.show_tags()}")
print(f"Versions: {model.show_versions()}")
```
