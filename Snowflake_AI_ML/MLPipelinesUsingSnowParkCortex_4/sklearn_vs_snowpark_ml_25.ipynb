```
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from snowflake.snowpark import Session
from xgboost import XGBClassifier
from snowflake.ml.modeling.xgboost import XGBClassifier as XGBC
import time
```

```
X,y=make_classification(n_samples=1000000, 
                        n_features=6,n_informative=4,
                        n_redundant=1,random_state=0,
                        shuffle=True)
```

```
X=pd.DataFrame(data=X,columns=['X1','X2','X3','X4','X5','X6'])
y=pd.DataFrame(data=y,columns=['y'])
pdf=pd.concat(objs=[X,y],axis=1)
```

```
session=Session.builder.getOrCreate()
```

```
df=session.create_dataframe(pdf)
```

```
df.show()
```

```
df.write.mode("overwrite").save_as_table("CLASSIFICATION_DATASET")
```

```
session.query_tag='classification'
```

```
df=session.table('CLASSIFICATION_DATASET').to_pandas()
df.head()
```

```
X=pd.DataFrame(data=df, columns=['X1','X2','X3','X4','X5','X6'])
y=pd.DataFrame(data=df, columns=['y'])
```

```
X.head()
```

```
y.head()
```

```
X_train,X_test, y_train,y_test=train_test_split(X,y, test_size=0.1)
```

```
start_time= time.time()

clf=XGBClassifier()

clf.fit(X_train,y_train)

end_time=time.time()

print(df.head())

print(f"Total time taken: {(end_time-start_time)} seconds")
```
